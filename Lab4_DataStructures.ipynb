```python
# Student Identity
LAST_NAME = "Mendoza"  # Replace with surname
STUDENT_ID = "TUPM-26-1250" # Replace with ID

# Derive computational parameters
seed_digit = int(STUDENT_ID[-1])
id_checksum = sum(int(d) for d in STUDENT_ID if d.isdigit())
vector_dim = len(LAST_NAME)

# Encapsulate state into a Dictionary (Associative Map)
sys_config = {
    "operator": LAST_NAME,
    "auth_id": STUDENT_ID,
    "base_seed": seed_digit,
    "checksum": id_checksum,
    "vector_dim": vector_dim,
    "status": "INITIALIZED"
}

# Display system state
print("=== SYSTEM CONFIGURATION ===")
for key, value in sys_config.items():
    print(f"{key.upper()}: {value}")
```

```python
# --- PHASE 3: MUTABLE SEQUENCES (LISTS) ---

# Task 5: Dynamic List Operations
base_val = sys_config["base_seed"]
number_sequence = [base_val, base_val + 15, sys_config["checksum"]]
print(f"Initial Sequence: {number_sequence}")

number_sequence.append(base_val + 20)
print(f"After Append: {number_sequence}")

# Task 6: Sequential Operations
new_numbers = [base_val + 5, sys_config["vector_dim"], base_val]
number_sequence.extend(new_numbers)
print(f"After Extend: {number_sequence}")

base_count = number_sequence.count(base_val)
print(f"Occurrences of {base_val}: {base_count}")

number_sequence.sort()
print(f"Sorted Sequence: {number_sequence}")
```

```python
# --- PHASE 4: IMMUTABLE RECORDS & ASSOCIATIVE MAPPING ---

# Task 7: Immutable Sequences (Tuples)
fixed_coordinates = (sys_config["vector_dim"], sys_config["base_seed"], 0)
print(f"Fixed Coordinates: {fixed_coordinates}")

# Unpack the tuple into individual variables
x_val, y_val, z_val = fixed_coordinates
print(f"Unpacked Values -> X: {x_val}, Y: {y_val}, Z: {z_val}")

# Attempt to modify the tuple (Expected to fail)
print("\nAttempting modification...")
try:
    fixed_coordinates[0] = 99
except TypeError as error_msg:
    print(f"Modification Error: {error_msg}")

# Task 8: Key-Value Mappings (Dictionaries)
data_payload = {
    "identifier": sys_config["auth_id"],
    "dimension": sys_config["vector_dim"],
    "status": "active"
}

# Iterate and display key-value pairs
print("\n--- Payload Data ---")
for key, value in data_payload.items():
    print(f"{key.capitalize()}: {value}")

# Add a new key-value pair dynamically
data_payload["efficiency_rating"] = 98.5
print(f"\nUpdated Payload: {data_payload}")
```

```python
# --- PHASE 5: UNIQUE COLLECTIONS (SETS) ---

# Task 9: Data De-duplication
base_val = sys_config["base_seed"]

# Initialize a list with deliberate duplicate values
raw_data = [base_val, base_val + 5, base_val, sys_config["vector_dim"], base_val + 5]
print(f"Raw List (with duplicates): {raw_data}")

# Convert the list to a Set to remove duplicates
unique_data = set(raw_data)
print(f"Unique Set: {unique_data}")

# Task 10: Mathematical Set Operations
# Define a secondary reference set for comparison
reference_set = {base_val, base_val + 10, base_val + 20}
print(f"Reference Set: {reference_set}")

# Perform mathematical intersections and unions
common_elements = unique_data.intersection(reference_set)
combined_elements = unique_data.union(reference_set)

print(f"Intersection (Common): {common_elements}")
print(f"Union (Combined): {combined_elements}")
```

```python
# --- PHASE 6: ADVANCED ITERATIONS ---
import array

# Task 11: Homogeneous Arrays
number_array = array.array('i', [sys_config["base_seed"], sys_config["vector_dim"], 100])
print(f"Integer Array: {number_array}")

# Attempt to append a non-integer data type (Expected to fail)
print("\nAttempting to append a string...")
try:
    number_array.append("invalid")
except TypeError as error_msg:
    print(f"Type Error: {error_msg}")

# Task 12: List Comprehensions
base_val = sys_config["base_seed"]

# Generate a sequence dynamically (multiplying by base_val)
generated_list = [(x * base_val) for x in range(1, 6)]
print(f"\nGenerated Comprehension: {generated_list}")

# Filter the generated sequence for values greater than 15
filtered_list = [x for x in generated_list if x > 15]
print(f"Filtered Comprehension (Values > 15): {filtered_list}")
```

```python
# --- PHASE 7: SPECIALIZED COLLECTIONS ---
from collections import deque, defaultdict

# Task 13: Double-Ended Queues (Deque)
data_queue = deque([sys_config["base_seed"], sys_config["vector_dim"]])
print(f"Initial Deque: {data_queue}")

# Append to both ends
data_queue.append(100)
data_queue.appendleft(200)
print(f"After Appends: {data_queue}")

# Remove from the left side
data_queue.popleft()
print(f"Final Deque (After Popleft): {data_queue}")

# Task 14: Fault-Tolerant Key Mappings (Defaultdict)
# Initialize with a default integer factory (creates a 0 for missing keys)
default_data = defaultdict(int)

# Assign a known key
default_data['active_key'] = sys_config["vector_dim"]
print(f"\nExisting Key Value: {default_data['active_key']}")

# Accessing an uninitialized key will not throw a KeyError
print(f"Missing Key Value (Auto-generated): {default_data['unknown_key']}")
```

```python
# === ADDITIONAL EXERCISE 1: Engineering Test Data Organizer ===
from collections import defaultdict

# Setup Data
LAST_NAME = "Mendoza"
SEED_NUM = 0
FAVORITE_ARTIST = "Syd Hartha"

# Generate numerical values using SEED_NUM and character positions (ignoring spaces)
generated_values = [SEED_NUM + i for i, char in enumerate(FAVORITE_ARTIST, 1) if char.strip()]

sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]
categories = ["TEMPERATURE", "VOLTAGE", "CURRENT", "PRESSURE", "SPEED"]

# 1. Generated Test Data (Includes multiple categories and a repeated record)
records = [
    (sources[0], categories[0], generated_values[0]),
    (sources[0], categories[0], generated_values[1]), # Multiple in one category
    (sources[1], categories[1], generated_values[2]),
    (sources[1], categories[1], generated_values[3]), # Multiple in another
    (sources[2], categories[2], generated_values[4]),
    (sources[3], categories[3], generated_values[5]),
]
records.append(records[5]) # Repeated test-source/category record

# 2. Organized Test Data
organized_data = defaultdict(lambda: defaultdict(list))
for src, cat, val in records:
    organized_data[src][cat].append(val)

# 3. Repeated/Distinct Data
unique_records = set(records)
repeated_records = [r for r in unique_records if records.count(r) > 1]

# 4 & 5. Numerical/Categorical Summaries and Processed Results
print("=== EXERCISE 1: ENGINEERING DATA RESULTS ===")
print(f"All Records: {records}\n")
print(f"Repeated Records Identified: {repeated_records}\n")
print("Categorical Summaries:")

category_summary = defaultdict(list)
for src, cat, val in records:
    category_summary[cat].append(val)

for cat, vals in category_summary.items():
    print(f" - {cat}: Count={len(vals)}, Max={max(vals)}, Min={min(vals)}")
```

```python
# === ADDITIONAL EXERCISE 2: Student Performance Data Analyzer ===

students = [f"{LAST_NAME}_S{i}" for i in range(1, 6)]
subjects = ["PROGRAMMING", "MATHEMATICS", "ELECTRONICS", "PHYSICS", "ENGLISH"]

# 1. Generate Performance Records
perf_records = [
    (students[0], subjects[0], generated_values[0]),
    (students[1], subjects[1], generated_values[1]),
    (students[2], subjects[2], generated_values[2]),
    (students[3], subjects[3], generated_values[3]),
]
perf_records.append((students[0], subjects[0], generated_values[0])) # Repeated record
perf_records.append((students[4], subjects[4], None)) # Incomplete record (missing score)

# 3. Performance Categories
def categorize_performance(score):
    if score is None: return "INCOMPLETE"
    return "SATISFACTORY" if score >= 5 else "NEEDS IMPROVEMENT"

# 4 & 5. Identify Records and Summarize
incomplete = [r for r in perf_records if r[2] is None]
repeated_perf = [r for r in set(perf_records) if perf_records.count(r) > 1]

print("=== EXERCISE 2: PERFORMANCE DATA RESULTS ===")
print(f"Incomplete Records Identified: {incomplete}")
print(f"Repeated Records Identified: {repeated_perf}\n")
print("Student Summaries (Categorized):")

for student, subject, score in set(perf_records): # Set used to filter out the exact duplicate for summary
    print(f" - {student} in {subject}: Score = {score} ({categorize_performance(score)})")
```

```python
# === ADDITIONAL EXERCISE 3: Transaction History Analyzer ===

customers = [f"{LAST_NAME}_C{i}" for i in range(1, 6)]
products = ["NOTEBOOK", "PEN", "USB", "HEADPHONES", "CALCULATOR"]
prices = {"NOTEBOOK": 50, "PEN": 15, "USB": 300, "HEADPHONES": 500, "CALCULATOR": 850}

# 1. Generate Transaction Records (Customer, Product, Quantity)
transactions = [
    (customers[0], products[0], generated_values[0]),
    (customers[1], products[1], generated_values[1]),
    (customers[2], products[2], generated_values[2]),
    (customers[3], products[3], generated_values[3]),
    (customers[4], products[4], generated_values[4]),
]
transactions.append((customers[0], products[0], generated_values[0])) # Repeated purchase

print("=== EXERCISE 3: TRANSACTION ANALYZER RESULTS ===")
total_revenue = 0
customer_totals = defaultdict(int)

# 3. Determine Transaction Values and Summaries
print("Detailed Transaction Log:")
for cust, prod, qty in transactions:
    transaction_value = qty * prices[prod]
    total_revenue += transaction_value
    customer_totals[cust] += transaction_value
    print(f" - {cust} purchased {qty} {prod}(s) for P{transaction_value}")

print("\nCustomer Spending Summary:")
for cust, total in customer_totals.items():
    print(f" - {cust} Total Spend: P{total}")

print(f"\nOverall Dataset Revenue: P{total_revenue}")
```
